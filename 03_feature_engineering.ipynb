```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

print("pandas:", pd.__version__)
print("numpy :", np.__version__)
```

```python
DATA_DIR = Path("clean")
DATA_DIR1 = Path("features")

FEATURE_DIR = DATA_DIR1 

CLEAN_FILE = DATA_DIR / "market_data_1h_clean.csv"
FEATURE_FILE = FEATURE_DIR / "market_features_1h.csv"

FEATURE_DIR.mkdir(parents=True, exist_ok=True)

print("Input :", CLEAN_FILE)
print("Output:", FEATURE_FILE)
```

```python
df = pd.read_csv(
    CLEAN_FILE,
    index_col=0,
    parse_dates=True
)

df = df.sort_index()

print("=" * 70)
print("CLEAN DATASET")
print("=" * 70)

print("Rows    :", len(df))
print("Columns :", len(df.columns))
print("Start   :", df.index.min())
print("End     :", df.index.max())

display(df.head())
```

```python
print("=" * 70)
print("DATA VALIDATION")
print("=" * 70)

print("Duplicate timestamps:", df.index.duplicated().sum())
print("Total NaN:", df.isna().sum().sum())

print("\nColumns:")
for col in df.columns:
    print(" -", col)
```
